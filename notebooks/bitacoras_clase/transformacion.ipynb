```python
from openpyxl import load_workbook

libro = load_workbook("../../data/BASE_DATOS_VIVEROS_MAYO_2026.xlsx")
```

```python
#libro.sheetnames devuelve una lista de cadenas de texto con los nombres de todas las pestañas (hojas) que existen dentro del archivo Excel.
print(libro.sheetnames)

#selecciona la oja del excel activa y la guarda en la variable hoja
hoja = libro.active

#hoja.merged_cells.ranges accede a una oclección interna de objetos de tipo CellRange este representa todos los bloques de celdas combinadas en la hoja seleccionada(hoja).

# el for es para recorrer todos los bloques en los que aparecen celdas combinadas , con el print(rango) se mprime en pantalla el rango exactoe n coordenadas excel donde están las celdas combinadas.
for rango in hoja.merged_cells.ranges:
    print(rango)
```

```python
df_copia = df.copy()
df_copia.head(5)
```

```python
for rango in hoja.merged_cells.ranges:
    if rango.min_col == rango.max_col:
        columna = rango.min_col - 1
        fila_inicio = rango.min_row -2
        fila_fin = rango.max_row -1

        if fila_inicio < 0 < fila_fin > len(df_copia):
            continue
        valor = hoja.cell(row=rango.min_row, column=rango.min_col).value

        df_copia.iloc[fila_inicio:fila_fin, columna] = valor

```

```python
df_copia.to_excel("relleno.xlsx",index=False)
```
